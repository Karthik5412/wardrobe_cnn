```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/drive/MyDrive/Deep_Learning/custumes/color_combinations_60000_rgb.csv')
df.head()
```

```python
unwanted = ["Shirt_Color","Pants_Color", "Shoes_Color", "Vibe"]

df = df.drop(columns=unwanted)
df = df.rename(columns={
    "Shirt_RGB_R" :"Shirt_Red",
    "Shirt_RGB_G" :"Shirt_Green",
    "Shirt_RGB_B" :"Shirt_Blue",
    "Pants_RGB_R" :"Pants_Red",
    "Pants_RGB_G" :"Pants_Green",
    "Pants_RGB_B" :"Pants_Blue",
    "Shoes_RGB_R" :"Shoes_Red",
    "Shoes_RGB_G" :"Shoes_Green",
    "Shoes_RGB_B" :"Shoes_Blue"
})
```

```python
df.head(2)
```

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
```

```python
scaled_df = scale.fit_transform(df.drop(columns='Match_Score'))
final_df = df['Match_Score']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test, y_train, y_test = train_test_split(scaled_df,val_df,test_size=0.2,random_state= 42)
```

```python
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras import Sequential
```

```python
ann = Sequential([
    Dense(128,activation='relu', input_shape=(x_train.shape[1],)),
    Dense(64,activation='relu'),
    Dropout(0.3),
    Dense(32, activation='swish'),
    Dropout(0.2),
    Dense(16, activation='swish'),
    Dense(1,activation='softplus')
])
```

```python
ann.compile(optimizer='adam',loss="mean_squared_error",metrics=['mae'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

stops = EarlyStopping(monitor='val_loss', patience=3,restore_best_weights=True)
```

```python
ann.fit(x_train,y_train,batch_size=64,validation_data=(x_test,y_test),validation_batch_size=64, epochs=50, verbose=1, callbacks=[stops])
```

```python
ann.save('ann.keras')
```
